```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, Layout
import ipywidgets as widgets

def simular_polarizacao(Ex_amp=1.0, Ey_amp=1.0, delta_fase=0.0):
    # Parâmetros da onda
    t = np.linspace(0, 2*np.pi, 500)
    delta = np.radians(delta_fase)

    # Equações de Maxwell para as componentes de campo
    x = Ex_amp * np.cos(t)
    y = Ey_amp * np.cos(t + delta)

    # Configuração do Gráfico (A "Janela" do Mathlet)
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.plot(x, y, color='#1f77b4', lw=3)

    # Desenha os eixos
    ax.axhline(0, color='black', lw=1)
    ax.axvline(0, color='black', lw=1)

    # Limites e Estética
    ax.set_xlim([-1.5, 1.5])
    ax.set_ylim([-1.5, 1.5])
    ax.set_aspect('equal')
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.set_title(f'Trajetória do Vetor Campo Elétrico $\\vec{{E}}$\nPhase Shift: {delta_fase}°', fontsize=14)
    ax.set_xlabel('$E_x$', fontsize=12)
    ax.set_ylabel('$E_y$', fontsize=12)

    # Identificação do estado (Lógica Educacional)
    if delta_fase % 180 == 0:
        estado = "Linear"
    elif delta_fase % 360 in [90, 270] and Ex_amp == Ey_amp:
        estado = "Circular"
    else:
        estado = "Elíptica"

    plt.text(-1.4, -1.4, f"Estado: Polarização {estado}", fontsize=12,
             bbox=dict(facecolor='white', alpha=0.8))

    plt.show()

# Criação da Interface "Mathlet"
interact(simular_polarizacao,
         Ex_amp=FloatSlider(min=0.1, max=1.0, step=0.1, value=1.0, description='Amp Ex:'),
         Ey_amp=FloatSlider(min=0.1, max=1.0, step=0.1, value=1.0, description='Amp Ey:'),
         delta_fase=FloatSlider(min=0, max=360, step=5, value=0, description='Fase Δφ°:'))
```
